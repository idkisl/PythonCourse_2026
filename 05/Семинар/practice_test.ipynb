Используя модуль `UnitTest`, напишите набор тестов для функции `bubble_sort`, приведённой ниже.

Обязательно проверьте следующие ситуации:

0. корректность сортировки обычного списка чисел;
1. пустой список;
2. список одинаковых элементов;
3. список строк;
4. список, содержащий ноль;
5. список с элементами разных типов — в этом случае функция должна вызывать исключение `TypeError`.

Исходная функция:

```python
def compare(a, b):
    return a > b


def bubble_sort(array):
    for i in range(len(array)):
        for j in range(0, len(array) - i - 1):
            if compare(array[j], array[j + 1]):
                temp = array[j]
                array[j] = array[j + 1]
                array[j + 1] = temp
```

При необходимости доработайте функцию так, чтобы её поведение соответствовало требованиям тестов.
