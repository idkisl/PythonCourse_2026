## Часть 3. Индексация, фильтрация и сортировка

Далее используем один учебный набор данных о сотрудниках. Имена и показатели вымышленные. В нём специально есть пропуски и полностью повторяющаяся строка. Генерация данных — подготовительный код: его подробно разбирать не требуется.

**Задание 3.1. Генерация учебных записей**

Сначала создадим основу таблицы. Каждая строка соответствует одному сотруднику.

```python
rng = np.random.default_rng(2026)
names = [
    "Анна", "Борис", "Виктор", "Галина", "Дарья", "Егор",
    "Жанна", "Захар", "Ирина", "Кирилл", "Лариса", "Максим",
    "Надежда", "Олег", "Полина", "Роман", "София", "Тимур",
    "Ульяна", "Фёдор", "Юлия", "Ярослав", "Мария", "Денис"
]
n = len(names)
employees = pd.DataFrame({
    "employee_id": np.arange(1001, 1001 + n),
    "name": names,
    "department": rng.choice(["IT", "HR", "Sales"], size=n),
    "city": rng.choice(["Москва", "Казань", "Тверь"], size=n),
    "age": rng.integers(22, 56, size=n).astype(float),
    "experience": rng.integers(0, 15, size=n),
    "salary": rng.integers(70, 161, size=n).astype(float) * 1000,
})
employees.head()
```

**Задание 3.2. Добавление ошибок в данные**

Смоделируем пропущенные значения и дубликат. Позже будем находить и исправлять их с помощью Pandas.

```python
employees = employees.iloc[:len(names)].copy()  # повторный запуск не добавит ещё один дубликат
employees.loc[[2, 17], "salary"] = np.nan
employees.loc[7, "department"] = None
employees.loc[13, "age"] = np.nan
employees.loc[len(employees)] = employees.loc[5].copy()
employees
```

**Задание 3.3. Выбор столбцов по именам**

Выведите только имя, отдел и зарплату сотрудников. Исходная таблица не изменится.

```python
employees[["name", "department", "salary"]].head()
```

**Задание 3.4. Выбор через loc**

Возьмите строки с метками `0`–`3` включительно и три выбранных столбца.

```python
employees.loc[0:3, ["name", "age", "salary"]]
```

**Задание 3.5. Выбор через iloc**

Теперь выберите первые четыре строки и первые три столбца по их номерам.

```python
employees.iloc[0:4, 0:3]
```

**Задание 3.6. Простой фильтр**

Сначала получите всех сотрудников отдела `IT` и посмотрите только на имя и зарплату.

```python
employees.head()
```

```python
employees.loc[employees["department"] == "IT", ["name", "salary"]]
```

**Задание 3.7. Два условия одновременно**

Усложните фильтр: сотрудники отдела `IT` в возрасте от 30 лет. Для логического И используйте `&` и круглые скобки.

```python
employees.loc[
    (employees["department"] == "IT") & (employees["age"] >= 30)
]
```

**Задание 3.8. Одно из нескольких значений**

Найдите сотрудников Москвы или Казани. Для проверки принадлежности множеству вариантов удобен `isin()`.

```python
employees.loc[
    (employees["city"] == "Москва") | (employees["city"] == "Казань")
].head()
```

```python
employees.loc[
    employees["city"].isin(["Москва", "Казань"])
].head()
```

**Задание 3.9. Сортировка по одному столбцу**

Найдите пять сотрудников с наибольшей известной зарплатой.

```python
employees.sort_values("salary", ascending=False)[["name", "salary"]].head(5)
```

**Задание 3.10. Сортировка по двум столбцам**

Сначала отсортируйте по отделу (по возрастанию), затем внутри каждого отдела по зарплате (по убыванию).

```python
employees.sort_values(
    by=["salary", "department"],
    ascending=[False, True]
).head(10)
```

**Задание 3.11. Добавление новых признаков**

Создайте копию таблицы. Добавьте годовую зарплату и отметку об опыте работы от пяти лет.

```python
work = employees.copy()
work["annual_salary"] = work["salary"] * 12
work["experienced"] = work["experience"] >= 5
work[["name", "experience", "salary", "annual_salary", "experienced"]].head(10)
```

**Задание 3.12. Переименование столбцов**

Измените имя столбца `name` на `employee_name` в копии таблицы.

```python
work = work.rename(columns={"name": "employee_name"})
work.head(3)
```

**Задание 3.13. Удаление столбца**

Удалите из копии столбец `experienced`. Обратите внимание: методы часто возвращают **новую** таблицу.

```python
work = work.drop(columns=["experienced"])
work.columns.tolist()
```

`groupby()` делит строки на группы по значениям столбца. Затем к каждой группе можно применить агрегирующую функцию: `mean`, `sum`, `min`, `max`, `count` или `agg`.

**Задание 6.4. Число записей и число непустых значений**

```python
print("Строк по отделам:")
display(df.groupby("department").size())
print("Известных зарплат:")
display(df.groupby("department")["salary"].count())
```

**Задание 6.5. Несколько статистик одновременно**

С помощью `agg()` найдите среднее, медиану, минимум и максимум зарплат в каждом отделе.

```python
valid.groupby("department")["salary"].agg(["mean", "count", "min", "max", "sum"])
```

**Задание 6.6. Именованные агрегаты**

Сделайте результат удобнее: присвойте вычисленным столбцам собственные имена и сохраните отдел как обычный столбец с `as_index=False`.

```python
summary = valid.groupby("department", as_index=False).agg(
    employees=("employee_id", "count"),
    mean_salary=("salary", "mean"),
    median_salary=("salary", "median"),
    max_salary=("salary", "max")
)
summary
```

Расположите отделы по средней зарплате от большей к меньшей.

```python
summary.sort_values("mean_salary", ascending=False)
```

```python
sat_selected = sat_clean.loc[sat_clean["Number of Test Takers"] >= 30].copy()
```

**Задание 7.11. Создать вычисляемый столбец**

В столбце `SAT Total Mean` вычислите сумму трёх средних баллов школы: по чтению, математике и письму.

Это **сумма средних по компонентам для школы**, а не результат отдельного ученика.

```python
sat_selected["SAT Total Mean"] = (
    sat_selected["Critical Reading Mean"]
    + sat_selected["Mathematics Mean"]
    + sat_selected["Writing Mean"]
)
```

**Задание 7.12. Найти школы с наибольшим суммарным средним баллом**

```python
sat_selected[["School Name", "Number of Test Takers", "SAT Total Mean"]].sort_values(
    "SAT Total Mean", ascending=False
).head(10)
```

**Задание 7.13. Добавить признак для группировки**

В идентификаторе школы `DBN` третий символ обозначает район Нью-Йорка: `M` — Манхэттен, `X` — Бронкс, `K` — Бруклин, `Q` — Куинс, `R` — Статен-Айленд.

Обращение `.str[2]` извлекает третий символ **из каждой строки** столбца (в Python нумерация начинается с нуля).

```python
sat_selected["Borough Code"] = sat_selected["DBN"].str[2]
```

**Задание 7.14. Посчитать число школ по районам**

Проверьте, сколько школ из отобранной выборки находится в каждом районе.

```python
sat_selected["Borough Code"].value_counts()
```

**Задание 7.15. Составить групповой отчёт**

Для каждого района определите число школ, средний суммарный балл школы и максимальный суммарный балл школы. Используйте `groupby()` и именованные агрегаты в `agg()`.

```python
sat_report = sat_selected.groupby("Borough Code", as_index=False).agg(
    schools=("DBN", "count"),
    mean_sat_total=("SAT Total Mean", "mean"),
    max_sat_total=("SAT Total Mean", "max")
)
sat_report
```

**Задание 7.16. Отсортировать отчёт**

Расположите районы от большего среднего суммарного балла школы к меньшему.

**Важно:** среднее по школам — это не среднее по всем ученикам: каждая школа участвует в расчёте с одинаковым весом, хотя число сдававших SAT различается.

```python
sat_report = sat_report.sort_values("mean_sat_total", ascending=False)
sat_report
```

**Задание 7.17. Сохранить итоговый отчёт**

Экспортируйте результат в `sat_2010_report.csv` без индекса строк.

```python
sat_report.to_csv("sat_2010_report.csv", index=False)
```
