# Семинар 9. Введение в pandas

`pandas` предоставляет две основные структуры данных:

- **`Series`** — одномерная последовательность значений с индексами;
- **`DataFrame`** — двумерная таблица, состоящая из столбцов `Series`.

[Документация pandas](https://pandas.pydata.org/docs/) · [Быстрое введение](https://pandas.pydata.org/docs/user_guide/10min.html)

## Подготовка данных

Запустите следующую ячейку. Она создаст переменную `employees` и CSV-файл `employees_pandas.csv`. Данные синтетические; интернет и дополнительные файлы не нужны. При повторном запуске получится тот же набор данных.

```python
import pandas as pd
import numpy as np

# Учебный набор данных создаётся внутри ноутбука.
# Все имена и показатели вымышленные; случайная генерация воспроизводима.
rng = np.random.default_rng(2026)
names = [
    "Анна", "Борис", "Виктор", "Галина", "Дарья", "Егор",
    "Жанна", "Захар", "Ирина", "Кирилл", "Лариса", "Максим",
    "Надежда", "Олег", "Полина", "Роман", "София", "Тимур",
    "Ульяна", "Фёдор", "Юлия", "Ярослав", "Мария", "Денис"
]
n = len(names)
employees = pd.DataFrame({
    "employee_id": np.arange(1001, 1001 + n),
    "name": names,
    "department": rng.choice(["IT", "HR", "Sales"], size=n),
    "city": rng.choice(["Москва", "Казань", "Тверь"], size=n),
    "age": rng.integers(22, 56, size=n).astype(float),
    "experience": rng.integers(0, 15, size=n),
    "salary": rng.integers(70, 161, size=n).astype(float) * 1000,
})
employees.loc[[2, 17], "salary"] = np.nan
employees.loc[7, "department"] = None
employees.loc[13, "age"] = np.nan
# Добавим одну полностью повторяющуюся строку для упражнения на дубликаты.
employees.loc[len(employees)] = employees.loc[5].copy()
employees.to_csv("employees_pandas.csv", index=False)
print("Файл employees_pandas.csv создан. Размер таблицы:", employees.shape)
display(employees.head())
```
