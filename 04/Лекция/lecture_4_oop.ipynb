# 1. Что такое класс

Класс — это описание объектов одного типа.

В классе можно определить:

- какие данные хранят объекты;
- какие действия они умеют выполнять.

Сам класс — это ещё не конкретный объект.  
Конкретный объект создаётся как **экземпляр класса**.

````python
class Student:
    pass


student_1 = Student()
student_2 = Student()

print(student_1)
print(student_2)
````

Здесь `Student` — класс, а `student_1` и `student_2` — два разных экземпляра этого класса.

# 2. Определение атрибутов

Внутри класса можно задать поля — данные, связанные с объектами.

Сначала рассмотрим поля, записанные непосредственно внутри класса.

````python
class Student:
    age = 19
    average_mark = 4.5


student_1 = Student()
student_2 = Student()

print(student_1.age)
print(student_2.age)
````

Если изменить такое поле у конкретного объекта, у него появится собственное значение.

````python
student_1.age = 21

print(student_1.age)
print(student_2.age)
````

# 3. Метод без `self`

Метод — это функция, объявленная внутри класса.

Попробуем определить метод без параметра `self`.

````python
class Student:
    def hello():
        print("Привет!")


Student.hello()
````

Такой метод можно вызвать через сам класс:

```python
Student.hello()
```

Но при вызове через объект Python автоматически передаёт сам объект первым аргументом:

```python
student.hello()
```

поэтому обычный метод экземпляра должен принимать этот объект.

````python
class Student:
    def hello():
        print("Привет!")


student = Student()

Student.hello()

# Раскомментируйте:
student.hello()
````

# 4. Что такое `self`

`self` — это ссылка на конкретный объект, для которого вызван метод.

Важно:

```python
student.hello()
```

по смыслу близко к записи

```python
Student.hello(student)
```

Именно поэтому первым параметром обычного метода записывают `self`.

````python
class Student:
    def hello(self):
        print("Привет!")
        print("Текущий объект:", self)


student = Student()
student.hello() # hello(student)

student1 = Student()
student1.hello()
#print("Переменная student:", student)
````

# 5. Методы с `self`: практические примеры

Через `self` метод получает доступ к данным конкретного объекта.

````python
class Student:
    age = 19

    def say_age(self):
        print(f"Мне {self.age} лет")


student_1 = Student()
student_2 = Student()

student_2.age = 25

print(student_1.age)
print(student_2.age)

student_1.say_age()
student_2.say_age()
````

Ещё один пример: объект хранит число, а методы изменяют его состояние.

````python
class Counter:
    value = 0

    def increase(self):
        self.value += 1

    def decrease(self):
        self.value -= 1

    def show(self):
        print("Значение:", self.value)


counter = Counter()

counter.increase()
counter.increase()
counter.decrease()
counter.show()
````

# 6. Конструктор

На практике объекту часто нужно сразу передать данные при создании.

Для этого используется специальный метод:

```python
__init__
```

Он вызывается автоматически при создании экземпляра.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def info(self):
        print(f"{self.name}, {self.age} лет")


student_1 = Student("Анна", 18)
# student_1.__init__("Анна", 18)

student_2 = Student("Дмитрий", 20)

student_1.info() # info(student_1)
student_2.info()
````

Теперь каждый экземпляр хранит собственные значения `name` и `age`.

# 7. Инкапсуляция и скрытие

Идея инкапсуляции состоит в том, чтобы объединить данные и методы работы с ними внутри одного объекта.

Также внутренние детали объекта стараются не изменять напрямую извне.

В Python атрибут с двумя подчёркиваниями:

```python
self.__value
```

подвергается механизму **name mangling**.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount("Иван", 100000000)

account.deposit(500)
print(account.get_balance())
print(account.owner)

print(account.__balance)

# Такой доступ не работает обычным способом:
# print(account.__balance)
````

Аналогично можно скрывать и методы.

````python
class Example:
    def public_method(self):
        print("Открытый метод")
        self.__private_method()

    def __private_method(self):
        print("Скрытый метод")


obj = Example()
# obj.public_method()

obj.__private_method()
````

Важно понимать: это не абсолютная защита данных, а механизм, который помогает отделять публичный интерфейс объекта от его внутренней реализации.

# 8. Магические методы

Методы вида

```python
__method__
```

называют специальными или магическими.

Они позволяют определить поведение объектов в стандартных операциях Python.

Некоторые часто используемые методы:

| Операция | Метод |
|---|---|
| `print(obj)` | `__str__` |
| служебное представление | `__repr__` |
| `a + b` | `__add__` |
| `a - b` | `__sub__` |
| `a * b` | `__mul__` |
| `b * a` | `__rmul__` |
| `a == b` | `__eq__` |
| `a < b` | `__lt__` |
| `len(obj)` | `__len__` |
| `obj[i]` | `__getitem__` |
| `obj()` | `__call__` |

Разберём магические методы на примере вектора.

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, number):
        return Vector(self.x * number, self.y * number)

    def __rmul__(self, number):
        return self * number
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __len__(self):
        return self.x ** 2 + self.y ** 2
    
    def __getitem__(self, id):
        if id == 0:
            return self.x
        elif id == 1:
            return self.y
        
    def __call__(self, *args):
        print(args)


v1 = Vector(1, 2)
v2 = Vector(3, 4)

v3 = v1 + v2 # v3 = v1.__add__(v2)
v4 = v1 * 2 # v1.__mul__(2)
v5 = 2 * v1 # v1.__rmul__(2)
print(v4(1, 2, 3.5))

lst = [v1, v2, v3]
print(lst)
````

# 9. Полиморфизм

Полиморфизм означает, что объекты разных классов могут использоваться единообразно, если поддерживают нужные операции.

Например, у разных классов может быть метод с одинаковым именем.

````python
class Student:
    def info(self):
        return "Я студент"


class Teacher:
    def info(self):
        return "Я преподаватель"


objects = [Student(), Teacher(), Student()]

for obj in objects:
    print(obj.info())
````

Циклу не нужно отдельно проверять тип каждого объекта.  
Он просто вызывает `info()`.

# 10. Утиная типизация

В Python часто важно не то, к какому именно классу принадлежит объект, а то, умеет ли он выполнять нужную операцию.

Если объект имеет нужный метод, его можно использовать.

````python
class Dog:
    def voice(self):
        return "Гав!"


class Cat:
    def voice(self):
        return "Мяу!"


class Robot:
    def voice(self):
        return "Бип!"


def make_sound(obj):
    print(obj.voice())


make_sound(Dog())
make_sound(Cat())
make_sound(Robot())
````

Функция `make_sound` ничего не знает о классах `Dog`, `Cat` и `Robot`.  
Она предполагает только наличие метода `voice()`.

# 11. Наследование

Наследование позволяет создать новый класс на основе существующего.

Дочерний класс получает методы и свойства базового класса.

````python
class Person:
    def hello(self):
        print("Привет!")


class Student(Person):
    def study(self):
        print("Учусь")


student = Student()

student.hello()
student.study()

p = Person()
p.hello()
p.study()
````

# 12. Переопределение методов

Дочерний класс может заменить поведение метода, унаследованного от базового класса.

````python
class Animal:
    def voice(self):
        return "Неизвестный звук"


class Dog(Animal):
    def voice(self):
        return "Гав!"


class Cat(Animal):
    def voice(self):
        return "Мяу! " + super().voice()


animals = [Animal(), Dog(), Cat()]

for animal in animals:
    print(animal.voice())
````

При наследовании удобно вызывать конструктор базового класса с помощью `super()`.

````python
class Person:
    def __init__(self, name, surname):
        self.name = name
        self.surname = surname


class Student(Person):
    def __init__(self, name, surname, course):
        super().__init__(name, surname)
        self.course = course


student = Student("Анна", "Петрова", 1)

print(student.name)
print(student.surname)
print(student.course)
````

# 13. Множественное наследование

Python позволяет наследоваться сразу от нескольких классов.

````python
class CanRun:
    def run(self):
        print("Бегу")


class CanSwim:
    def swim(self):
        print("Плыву")


class Duck(CanRun, CanSwim):
    pass


duck = Duck()
duck.run()
duck.swim()
````

Если в родительских классах есть методы с одинаковым именем, Python использует порядок разрешения методов — **MRO**.

````python
class A:
    def f(self):
        print("A")


class B:
    def f(self):
        print("B")


class C(B, A):
    pass


obj = C()
obj.f()

print(C.mro())
````

# 14. Ромбовидное наследование

Ромбовидное наследование возникает, когда два класса наследуются от одного общего предка, а затем третий класс наследуется от них обоих.

Схема:

```text
      A
     / \
    B   C
     \ /
      D
```

````python
class A:
    def hello(self):
        print("A")


class B(A):
    pass


class C(A):
    def hello(self):
        print("C")


class D(B, C):
    pass


obj = D()

obj.hello()
print(D.mro())
````

MRO определяет, в каком порядке Python ищет нужный метод.

# 15. Наследование от встроенных классов

Можно наследоваться не только от собственных классов, но и от встроенных типов Python.

````python
class StrangeInt(int):
    def __add__(self, other):
        return self * other


a = StrangeInt(5)
b = StrangeInt(7)

print(a + b)
print(int(5) + int(7))
````

Ещё пример: список, который запрещает добавлять отрицательные числа.

````python
class PositiveList(list):
    def append(self, value):
        if value < 0:
            print("Можно добавлять только неотрицательные числа")
        else:
            super().append(value)


numbers = PositiveList()

numbers.append(5)
numbers.append(10)

print(numbers)

numbers.append(-3)
print(numbers)
````

### 2A. Пример с решением

Создайте класс `BankAccount`.

Баланс должен быть скрытым полем.

Добавьте методы:

- `deposit(amount)`;
- `withdraw(amount)`;
- `get_balance()`.

Не позволяйте снимать больше денег, чем находится на счёте.

````python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return True
        return False

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)

account.deposit(500)
account.withdraw(300)

print(account.get_balance())
````
