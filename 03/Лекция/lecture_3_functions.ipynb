## 4. `return`: функция как вычисление

`print()` и `return` решают разные задачи.

- `print()` показывает значение пользователю.
- `return` возвращает результат в программу.

````python
def square_print(x):
    print(x ** 2)

def square_return(x):
    return x ** 2

a = square_print(5)
b = square_return(5)

print("a =", a)
print("b =", b)
````

Если функция завершилась без `return`, Python неявно возвращает `None`.

## 5. Области видимости

Переменные, созданные внутри функции, обычно существуют только внутри неё.

````python
def demo():
    local_value = 100
    print("Внутри функции:", local_value)

demo()

# print(local_value)  # NameError
````

Переменная, созданная вне функции, является глобальной. Функция может читать её значение.

Использовать `global` для изменения глобального состояния можно, но обычно это делает код сложнее для анализа.

**Практическое правило:** если можно вернуть новое значение через `return`, это обычно лучше, чем использовать `global`.

````python
PASSWORD = "1234"

def pin_code(password):
    return password == PASSWORD

print(pin_code("1234"))
````

Если заранее неизвестно количество позиционных аргументов, их можно собрать в кортеж.

````python
def show_all(*args):
    print(args)
    print(type(args))

show_all(1, 2, 3, 4, 5)
````

# 10. Рекурсия

**Рекурсия** — ситуация, когда функция вызывает саму себя.

Чтобы рекурсивная функция завершилась, нужны:
1. **базовый случай** — когда дальнейший вызов не нужен;
2. **рекурсивный переход** — сведение задачи к более простой.

### Первый пример

Пусть

\[
F(20)=100,
\]

\[
F(N)=F(N+1)\cdot(N-1).
\]

Определение само подсказывает структуру программы.

````python
def F(n):
    if n == 20:
        return 100
    return F(n + 1) * (n - 1)

print(F(19))
````

### Наводящая задача 11

$F(N) = F(N - 1) + 3$.

Вычислите вручную `F(4)`, а затем реализуйте функцию.

````python
def F(n):
    if n == 1:
        return 1
    return F(n - 1) + 3

print(F(4))
````

### Второй пример

$F(0)=1,\qquad F(1)=2,$


$F(N)=N\cdot F(N-1)+F(N-2)+4N$

Здесь базовых случаев уже два.

````python
def F(n):
    if n == 0:
        return 1
    if n == 1:
        return 2
    return n * F(n - 1) + F(n - 2) + 4 * n

print(F(11))
````

### Факториал

$
n! = 1\cdot2\cdot3\cdots n
$

и одновременно

$
n! = n\cdot(n-1)!
$

Это естественная рекурсивная формула.

````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

# 11. Генераторы

`range(10)` не создаёт обычный список `[0, 1, ..., 9]`. Он выдаёт значения по мере необходимости.

Сначала реализуем похожую функцию через список.

````python
def my_range_list(n):
    result = []
    i = 0
    while i < n:
        result.append(i)
        i += 1
    return result

print(my_range_list(10))
````

Для большого `n` весь список приходится хранить в памяти. Генератор позволяет выдавать элементы по одному с помощью `yield`.

````python
def my_range(n):
    i = 0
    while i < n:
        yield i
        i += 1

for value in my_range(10):
    print(value, end=" ")
````

### `return` и `yield`

`return`:
- возвращает значение;
- завершает функцию.

`yield`:
- отдаёт очередное значение;
- приостанавливает функцию;
- при следующем обращении продолжает с того же места.

# 14. `map` и `filter`

`map(func, collection)` применяет функцию к каждому элементу.

`filter(criteria, collection)` оставляет только элементы, для которых функция-критерий возвращает `True`.

````python
numbers = list(range(10))

squares = map(lambda x: x ** 2, numbers)
print(list(squares))

even_numbers = filter(lambda x: x % 2 == 0, numbers)
print(list(even_numbers))
````

````python
def my_map(func, collection):
    for elem in collection:
        yield func(elem)

def my_filter(criteria, collection):
    for elem in collection:
        if criteria(elem):
            yield elem

print(list(my_map(lambda x: x ** 2, range(5))))
print(list(my_filter(lambda x: x > 2, range(5))))
````

# 15. Декораторы

Начнём не с синтаксиса `@`, а с задачи.

Пусть есть функция, и мы хотим добавлять отладочную информацию перед её выполнением, не меняя исходное тело.

````python
def output(a, b, c):
    print(a + b + c)
    return a * b * c

print(output(2, 3, 4))
````

````python
def new_func(a, b, c):
    print("INFO:", a, b, c)
    result = output(a, b, c)
    return result

print(new_func(2, 3, 4))
````

Теперь сделаем функцию, которая умеет оборачивать другую функцию.

````python
def decorate(func):
    def new_func(a, b, c):
        print("INFO:", a, b, c)
        result = func(a, b, c)
        return result
    return new_func

decorated_output = decorate(output)
print(decorated_output(2, 3, 4))
````

Запись с `@decorate` означает по сути:

```python
output = decorate(output)
```

````python
@decorate
def output(a, b, c):
    print(a + b + c)
    return a * b * c

print(output(2, 3, 4))
````

### Универсальный декоратор

Чтобы обёртка принимала функции с разным количеством аргументов, удобно использовать `*args` и `**kwargs`.

````python
def debug(func):
    def wrapper(*args, **kwargs):
        print("Вызов функции:", func.__name__)
        print("args:", args)
        print("kwargs:", kwargs)
        result = func(*args, **kwargs)
        print("Результат:", result)
        return result
    return wrapper

@debug
def add(a, b):
    return a + b

print(add(4, 7))
````
