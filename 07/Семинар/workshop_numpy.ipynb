# Шпаргалка по NumPy

Ниже собраны основные конструкции, которые встречались в лекции.

| Тема | Конструкция | Что делает |
|---|---|---|
| Создание массива | `np.array(data)` | Создаёт массив |
| Тип данных | `a.dtype` | Показывает тип элементов |
| Изменение типа | `a.astype(type)` | Создаёт массив с другим типом |
| Размерность | `a.ndim` | Количество измерений |
| Форма | `a.shape` | Размеры массива по осям |
| Число элементов | `a.size` | Общее число элементов |
| Изменение формы | `a.reshape(...)` | Меняет форму массива |
| В одномерный массив | `a.reshape(-1)` | Превращает массив в одномерный |
| В одномерный массив | `a.flatten()` | Возвращает одномерную копию |
| В одномерный массив | `a.ravel()` | Разворачивает массив, по возможности без копирования |
| Копия | `a.copy()` | Создаёт независимую копию |
| Нули | `np.zeros(shape)` | Создаёт массив нулей |
| Единицы | `np.ones(shape)` | Создаёт массив единиц |
| Диапазон | `np.arange(start, stop, step)` | Аналог `range` для массива |
| Равномерная сетка | `np.linspace(start, stop, n)` | Создаёт `n` равноотстоящих чисел |
| Единичная матрица | `np.eye(n)` | Создаёт единичную матрицу |
| Генератор случайных чисел | `np.random.default_rng(seed)` | Создаёт генератор |
| Случайные целые | `rng.integers(a, b, size=...)` | Генерирует целые числа |
| Случайные вещественные | `rng.random(size=...)` | Числа из диапазона `[0, 1)` |
| Нормальные случайные числа | `rng.normal(...)` | Генерирует числа нормального распределения |
| Индексация | `a[i]`, `a[i, j]` | Получает отдельный элемент |
| Срез | `a[start:stop:step]` | Получает часть массива |
| Строка матрицы | `a[i, :]` | Выбирает строку |
| Столбец матрицы | `a[:, j]` | Выбирает столбец |
| Векторизация | `a + b`, `a * b`, `a ** 2` | Поэлементные операции |
| Broadcasting | `a + b` при совместимых `shape` | Расширяет массивы при вычислениях |
| Новая ось | `a[:, None]`, `a[None, :]` | Превращает вектор в столбец/строку |
| Маска | `a[a > 0]` | Отбирает элементы по условию |
| Несколько условий | `(a > 0) & (a < 10)` | Логическое И |
| Несколько условий | `(a < 0) \| (a > 10)` | Логическое ИЛИ |
| Отрицание условия | `~mask` | Логическое НЕ |
| Выбор по условию | `np.where(cond, x, y)` | Векторный аналог `if` |
| Есть хотя бы один | `np.any(cond)` | Проверяет, есть ли `True` |
| Выполняется для всех | `np.all(cond)` | Проверяет, все ли `True` |
| Уникальные значения | `np.unique(a)` | Возвращает уникальные элементы |
| Частоты | `np.unique(a, return_counts=True)` | Уникальные элементы и их количества |
| Сумма | `a.sum()` / `np.sum(a)` | Сумма элементов |
| Произведение | `a.prod()` | Произведение элементов |
| Минимум | `a.min()` | Минимальное значение |
| Максимум | `a.max()` | Максимальное значение |
| Среднее | `a.mean()` | Среднее арифметическое |
| Стандартное отклонение | `a.std()` | Стандартное отклонение |
| Индекс минимума | `a.argmin()` | Позиция минимума |
| Индекс максимума | `a.argmax()` | Позиция максимума |
| Медиана | `np.median(a)` | Медиана |
| Процентиль | `np.percentile(a, p)` | `p`-й процентиль |
| Ненулевые элементы | `np.count_nonzero(a)` | Число ненулевых элементов |
| Операция по столбцам | `func(a, axis=0)` | Агрегирует строки |
| Операция по строкам | `func(a, axis=1)` | Агрегирует столбцы |
| Сохранение оси | `keepdims=True` | Не удаляет агрегируемую ось |
| Корень | `np.sqrt(a)` | Квадратный корень |
| Логарифм | `np.log(a)` | Натуральный логарифм |
| Экспонента | `np.exp(a)` | Экспонента |
| Синус | `np.sin(a)` | Синус |
| Косинус | `np.cos(a)` | Косинус |
| Проверка `NaN` | `np.isnan(a)` | Находит `NaN` |
| Проверка бесконечности | `np.isinf(a)` | Находит `±inf` |
| Проверка корректности | `np.isfinite(a)` | Находит конечные значения |
| Вертикальное объединение | `np.vstack((a, b))` | Склеивает по строкам |
| Горизонтальное объединение | `np.hstack((a, b))` | Склеивает по столбцам |
| Общее объединение | `np.concatenate((a, b), axis=...)` | Объединяет по выбранной оси |
| Вставка | `np.insert(...)` | Вставляет элементы |
| Удаление | `np.delete(...)` | Удаляет элементы |
| Добавление | `np.append(...)` | Добавляет элементы |
| Индексы по условию | `np.where(cond)` | Возвращает позиции |
| Индексы ненулевых | `np.nonzero(a)` | Возвращает позиции ненулевых элементов |
| Сортировка | `np.sort(a)` | Возвращает отсортированную копию |
| Индексы сортировки | `np.argsort(a)` | Возвращает порядок индексов |
| Транспонирование | `a.T` | Меняет строки и столбцы местами |
| Поэлементное умножение | `A * B` | Умножает элементы |
| Матричное умножение | `A @ B` | Матричное произведение |
| Определитель | `np.linalg.det(A)` | Вычисляет определитель |
| Обратная матрица | `np.linalg.inv(A)` | Вычисляет обратную матрицу |
| СЛАУ | `np.linalg.solve(A, b)` | Решает `Ax = b` |
