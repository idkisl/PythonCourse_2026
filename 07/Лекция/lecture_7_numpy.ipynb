## 1. Зачем вообще нужен NumPy?

Python удобен, но обычный цикл выполняет много операций на уровне интерпретатора.

Если нам нужно обработать **миллионы чисел**, то удобнее передать вычисление библиотеке, внутри которой основные операции реализованы на низком уровне.

Главная идея NumPy:

> **Не обрабатывай каждый элемент вручную, если можно описать операцию над всем массивом сразу.**


### Задание 1

Вспомните достоинства и недостатки Python.

Обсудите:

1. Почему Python удобен для разработки?
2. Почему обычные циклы Python могут быть медленными?
3. Что значит «векторизовать» вычисление?


````python
import timeit

python_time = timeit.timeit(
    "[i ** 0.5 for i in range(1, 1_000_000)]",
    number=3
)

numpy_time = timeit.timeit(
    "__import__('numpy').arange(1, 1_000_000) ** 0.5",
    number=3
)

print("Python:", python_time)
print("NumPy :", numpy_time)
````

**Важно:** сравнение времени — это не доказательство того, что NumPy всегда быстрее.  
Но для больших числовых массивов векторизованные операции NumPy обычно значительно эффективнее обычных циклов Python.


# 2. Массив `ndarray`

Основной объект NumPy — `ndarray`.

Список Python и массив NumPy похожи внешне, но устроены по-разному.

### Список Python

```python
[1, 2, 3]
```

может содержать объекты разных типов.

### Массив NumPy

Обычно хранит элементы **одного типа**, благодаря чему библиотека может компактно размещать данные в памяти и быстро выполнять над ними одинаковые операции.


````python
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr)
print(type(arr))
````

### `dtype` — тип элементов массива

NumPy самостоятельно выбирает подходящий тип, но его можно указать явно.


````python
a = np.array([1, 2, 3])
b = np.array([1, 2, 3], dtype=np.float64)
c = np.array([1, 2, 3], dtype=np.int8)

print(a, a.dtype)
print(b, b.dtype)
print(c, c.dtype)
````

Если в массив поместить данные разных типов, NumPy попытается привести их к единому типу.


````python
mixed = np.array([1, 2, 3, "hello"])
print(mixed)
print(mixed.dtype)
````

### Преобразование типа: `astype`

Тип массива можно изменить после создания.

```python
new_array = old_array.astype(new_type)
```

`astype` создаёт новый массив с другим `dtype`.


````python
a = np.array([1.2, 2.8, 3.5])

print(a)
print(a.dtype)

b = a.astype(int)

print(b)
print(b.dtype)
````

Обратите внимание: преобразование `float → int` **не округляет** число, а отбрасывает дробную часть.

Если нужно именно округление, сначала используйте, например, `np.round`.


````python
a = np.array([1.2, 2.8, 3.5])

print(a.astype(int))
print(np.round(a).astype(int))
````

# 3. Размерность массива: `ndim`, `shape`, `size`

У массива есть несколько важных характеристик:

- `ndim` — количество измерений;

## Изменение формы: `reshape`

`reshape` меняет представление данных, но количество элементов должно сохраниться.


````python
arr = np.arange(1, 9)

print(arr.reshape(2, 4))
print()
print(arr.reshape(4, 2))
print()
print(arr.reshape(2, 2, 2))
````

Можно использовать `-1`, если один из размеров должен быть вычислен автоматически.


````python
arr = np.arange(12)

print(arr.reshape(3, -1))
print(arr.reshape(-1, 2))
````

- `reshape(-1)`;
- `flatten()`;
- `ravel()`.


````python
matrix = np.arange(12).reshape(3, 4)

print(matrix)
print()
print(matrix.reshape(-1))
print(matrix.flatten())
print(matrix.ravel())
````

На первом знакомстве достаточно считать их способами «развернуть» массив в строку.

Позже полезно помнить: `flatten()` всегда создаёт копию, а `ravel()` по возможности старается вернуть представление исходных данных.

# 4.1. Важная особенность срезов: view и copy

У обычного списка Python срез создаёт новый список.

У NumPy срез часто является **представлением** (`view`) исходного массива.

Это значит, что изменение среза может изменить исходный массив.


````python
a = np.array([10, 20, 30, 40, 50])

part = a[1:4]
part[0] = 999

print("part =", part)
print("a    =", a)
````

Если нужна независимая копия, используйте `.copy()`.


````python
a = np.array([10, 20, 30, 40, 50])

part = a[1:4].copy()
part[0] = 999

print("part =", part)
print("a    =", a)
````

### Почему это важно?

При работе с большими массивами копирование данных может быть дорогим по памяти и времени.

Поэтому NumPy часто старается не копировать данные без необходимости.

Но программисту важно понимать, изменяет он отдельный массив или представление исходного.

# 7. Broadcasting

Иногда формы массивов различаются, но NumPy всё равно может выполнить операцию.

Это называется **broadcasting**.

Простейший пример — массив плюс число:


````python
matrix = np.arange(1, 7).reshape(2, 3)
print(matrix)
print()
print(matrix + 100)
````

Можно прибавить к каждой строке один и тот же вектор:


````python
matrix = np.arange(1, 7).reshape(2, 3)
row = np.array([10, 20, 30])

print(matrix)
print("+")
print(row)
print("=")
print(matrix + row)
````

### Главное правило на первом знакомстве

NumPy сравнивает размеры массивов **справа налево**.

Размеры совместимы, если они:

- одинаковы;
- либо один из них равен `1`.

Например:

- `(2, 3)` и `(3,)` — совместимы;
- `(5, 2)` и `(2,)` — совместимы;
- `(2, 5)` и `(2,)` — обычно **не совместимы**.


````python
a = np.arange(10).reshape(5, 2)
b = np.array([100, 200])

print(a + b)
````

## Форма вектора имеет значение

На первый взгляд массивы

```python
[1, 2, 3]
```

```python
[[1, 2, 3]]
```

```python
[[1],
 [2],
 [3]]
```

содержат одни и те же числа.

Но их формы различаются.


````python
a = np.array([1, 2, 3])
row = np.array([[1, 2, 3]])
column = np.array([[1], [2], [3]])

print("a      :", a.shape)
print("row    :", row.shape)
print("column :", column.shape)
````

Получаем:

- `(3,)` — одномерный массив;
- `(1, 3)` — одна строка, три столбца;
- `(3, 1)` — три строки, один столбец.

Это особенно важно при broadcasting.


````python
row = np.array([[10, 20, 30]])
column = np.array([[1], [2], [3]])

print(row + column)
````

Здесь формы `(1, 3)` и `(3, 1)` расширяются до `(3, 3)`.

Получается таблица всех возможных сумм.

# 8. Сравнения и булевы маски

Сравнение массива с числом возвращает массив `True` / `False`.


````python
arr = np.array([3, 8, 1, 10, 6])

mask = arr > 5

print(mask)
print(arr[mask])
````

Условие можно записывать сразу внутри индексации:


````python
arr = np.array([3, 8, 1, 10, 6])

print(arr[arr > 5])
````

Для нескольких условий используйте:

- `&` — И;
- `|` — ИЛИ;
- `~` — НЕ.

Каждое сравнение лучше заключать в скобки.


````python
arr = np.arange(1, 21)

result = arr[(arr > 5) & (arr % 2 == 0)]
print(result)
````

Маску можно использовать не только для чтения, но и для изменения элементов.


````python
data = np.array([-3, 5, -1, 7, 0, -8])
data[data < 0] = 0

print(data)
````

## 8.1. `np.where`

`np.where` позволяет выбирать одно значение, если условие истинно, и другое — если ложно.

По смыслу это похоже на векторный `if`.


````python
a = np.array([-5, 3, -2, 8, 0, 4])

result = np.where(a > 0, a, 0)

print(result)
````

Например, можно заменить все отрицательные значения на `-1`:


````python
a = np.array([-5, 3, -2, 8, 0, 4])

print(np.where(a < 0, -1, a))
````

Или создать массив подписей:


````python
scores = np.array([45, 81, 59, 92])

labels = np.where(scores >= 60, "зачёт", "незачёт")
print(labels)
````

## 8.2. `any` и `all`

Иногда нужно проверить не отдельные элементы, а условие для всего массива.

- `np.any(condition)` — существует ли хотя бы один подходящий элемент;
- `np.all(condition)` — подходят ли все элементы.


````python
a = np.array([4, 7, 9, 12])

print(np.any(a > 10))
print(np.all(a > 0))
print(np.all(a % 2 == 0))
````

Это особенно удобно для проверки входных данных.


````python
ages = np.array([16, 17, 18, 16])

if np.all(ages >= 16):
    print("Все подходят по возрасту")
````

# 9. Агрегатные функции

NumPy умеет быстро получать общие характеристики массива.


````python
arr = np.array([4, 8, 1, 10, 7])

print("sum :", arr.sum())
print("prod:", arr.prod())
print("min :", arr.min())
print("max :", arr.max())
print("mean:", arr.mean())
print("std :", arr.std())
````

## Дополнительные агрегаты

Кроме суммы, минимума и среднего, часто используют:

- `np.median` — медиана;
- `np.percentile` — процентиль;
- `np.count_nonzero` — количество ненулевых элементов.


````python
a = np.array([1, 2, 3, 4, 100])

print("mean   =", np.mean(a))
print("median =", np.median(a))
print("75%    =", np.percentile(a, 75))
print("nonzero=", np.count_nonzero(a))
````

Медиана особенно полезна, когда в данных есть выбросы.

В примере выше число `100` сильно увеличивает среднее, но гораздо меньше влияет на медиану.


# 10. Что такое `axis`?

Для двумерного массива можно выполнять операцию:

- над **всем массивом**;
- отдельно по строкам;
- отдельно по столбцам.

Рассмотрим матрицу:


````python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
````

### `axis=0`

Мы «убираем строки» и получаем результат для каждого столбца.


````python
print(matrix.sum(axis=0))
````

### `axis=1`

Мы «убираем столбцы» и получаем результат для каждой строки.


````python
print(matrix.sum(axis=1))
````

Удобная мысль:

> номер `axis` показывает, **какая ось исчезнет** после агрегирования.


### Сохранение размерности: `keepdims=True`

Иногда после агрегирования удобно не удалять ось.


````python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix.mean(axis=1))
print(matrix.mean(axis=1, keepdims=True))
````

Это особенно полезно, если результат затем участвует в broadcasting.

Например, вычтем из каждой строки её среднее:


````python
matrix = np.array([
    [1, 2, 3],
    [10, 20, 30]
], dtype=float)

row_means = matrix.mean(axis=1, keepdims=True)

print(matrix - row_means)
````

# 11. Математические функции NumPy

Функции NumPy умеют работать сразу со всем массивом.


````python
x = np.array([1, 4, 9, 16])

print(np.sqrt(x))
print(np.log(x))
print(np.exp([0, 1, 2]))
````

# 12. `inf` и `nan`

В вычислениях можно встретить специальные значения:

- `np.inf` — бесконечность;
- `-np.inf` — минус бесконечность;
- `np.nan` — `Not a Number`.


````python
print(np.inf)
print(-np.inf)
print(np.nan)
````

Некоторые операции приводят к `nan`.

NumPy обычно выдаёт предупреждение, но продолжает выполнение программы.


````python
with np.errstate(all="ignore"):
    print(np.sqrt(-1))
    print(np.inf - np.inf)
    print(0 * np.inf)
````

Проверять такие значения нужно специальными функциями:


````python
x = np.array([1.0, np.nan, np.inf, -np.inf, 5.0])

print(np.isnan(x))
print(np.isinf(x))
print(np.isfinite(x))
````

# 13. Объединение и изменение массивов

Массивы можно соединять по разным направлениям.


````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("vstack:")
print(np.vstack((a, b)))

print("\nhstack:")
print(np.hstack((a, b)))
````

Для более общего случая используется `concatenate`.


````python
print(np.concatenate((a, b), axis=0))
print()
print(np.concatenate((a, b), axis=1))
````

Также существуют `insert`, `delete`, `append`, но при частых изменениях размера массива обычный список Python может быть удобнее.

NumPy особенно хорош тогда, когда размер данных известен и над массивом выполняется много вычислений.


# 13.1. Поиск индексов и значений

Иногда нужны не сами значения, а позиции подходящих элементов.

Для этого можно использовать `np.where(condition)` или `np.nonzero`.


````python
a = np.array([3, 8, 1, 10, 6])

indices = np.where(a > 5)

print(indices)
print(a[indices])
````

Для простого одномерного массива часто удобно взять первый элемент результата:


````python
indices = np.where(a > 5)[0]
print(indices)
````

# 14. Сортировка

`np.sort` возвращает отсортированную копию.


````python
arr = np.array([8, 2, 10, 1, 5])
print(np.sort(arr))
````

В двумерном массиве сортировка зависит от `axis`.


````python
matrix = np.array([
    [8, 2, 5],
    [3, 9, 1]
])

print("axis=0:")
print(np.sort(matrix, axis=0))

print("\naxis=1:")
print(np.sort(matrix, axis=1))
````

Для получения **индексов** элементов в порядке сортировки используется `argsort`.


````python
arr = np.array([40, 10, 30, 20])

print(np.argsort(arr))
print(arr[np.argsort(arr)])
````

# 15. Матрицы

В NumPy двумерный массив может использоваться как матрица.

### Транспонирование


````python
A = np.arange(1, 7).reshape(2, 3)

print(A)
print()
print(A.T)
````

### Поэлементное и матричное умножение — разные операции

`A * B` — поэлементное умножение.

`A @ B` — матричное умножение.


````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A * B:")
print(A * B)

print("\nA @ B:")
print(A @ B)
````

Некоторые функции линейной алгебры находятся в `np.linalg`.


````python
A = np.array([
    [1, 2],
    [3, 4]
])

print("det =", np.linalg.det(A))
print("inverse:")
print(np.linalg.inv(A))
````

### Решение системы линейных уравнений

Система

\[
Ax=b
\]

может быть решена через `np.linalg.solve`.


````python
A = np.array([
    [2, 1],
    [1, -1]
], dtype=float)

b = np.array([5, 1], dtype=float)

x = np.linalg.solve(A, b)

print(x)
print(A @ x)
````
